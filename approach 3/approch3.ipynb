```
from datasets import load_dataset

# Login using e.g. `huggingface-cli login` to access this dataset
ds = load_dataset("Exploration-Lab/IL-TUR", "bail")
```

```
print(ds)
```

```
from datasets import concatenate_datasets

# Merge train
train_data = concatenate_datasets([
    ds["train_all"],
    ds["train_specific"]
])

# Merge dev
dev_data = concatenate_datasets([
    ds["dev_all"],
    ds["dev_specific"]
])

# Merge test
test_data = concatenate_datasets([
    ds["test_all"],
    ds["test_specific"]
])
```

```
print(train_data)
print(dev_data)
print(test_data)
```

```
train_data = train_data.shuffle(seed=42)
dev_data = dev_data.shuffle(seed=42)
test_data = test_data.shuffle(seed=42)
```

```
import pandas as pd

train_df = pd.DataFrame(train_data)
dev_df = pd.DataFrame(dev_data)
test_df = pd.DataFrame(test_data)
```

```
train_data = train_data.select(range(50000))
dev_data = dev_data.select(range(5000))
test_data = test_data.select(range(7000))
```

```
def format_example(example):
    return {
        "text": f"""<s>[INST] You are an expert Indian legal assistant.

CASE:
{example['text']}

Step-by-step:
1. Understand facts
2. Apply law
3. Give reasoning

Final Decision: [/INST]

{example['label']}</s>"""
    }

train_dataset = train_data.map(format_example)
dev_dataset = dev_data.map(format_example)
```

```
train_dataset = train_dataset.remove_columns(
    [col for col in train_dataset.column_names if col != "text"]
)

dev_dataset = dev_dataset.remove_columns(
    [col for col in dev_dataset.column_names if col != "text"]
)
```

```
import torch
from transformers import AutoTokenizer, AutoModelForCausalLM, BitsAndBytesConfig, AutoConfig

MODEL_NAME = "microsoft/Phi-3-mini-4k-instruct"

# Load config to inspect and potentially modify
config = AutoConfig.from_pretrained(MODEL_NAME, trust_remote_code=True)

# Check if 'rope_scaling' is defined and has 'type', if not, set a default
if not hasattr(config, 'rope_scaling') or config.rope_scaling is None or 'type' not in config.rope_scaling:
    print("rope_scaling not correctly defined in config. Setting a default to 'longrope'.")
    # Set to 'longrope' with required factors
    config.rope_scaling = {'type': 'longrope', 'factor': 1.0, 'short_factor': 1.0, 'long_factor': 1.0}

# Tokenizer
tokenizer = AutoTokenizer.from_pretrained(
    MODEL_NAME,
    trust_remote_code=True
)
tokenizer.pad_token = tokenizer.eos_token
```

```
bnb_config = BitsAndBytesConfig(
    load_in_4bit=True,
    bnb_4bit_compute_dtype=torch.float16,
    bnb_4bit_quant_type="nf4"
)
```

```
!pip install -U bitsandbytes
!pip install -U transformers accelerate peft trl
```

```
model = AutoModelForCausalLM.from_pretrained(
    MODEL_NAME,
    config=config, # Pass the pre-loaded and potentially modified config
    quantization_config=bnb_config,
    device_map="auto",
    trust_remote_code=True,
    dtype=torch.float16 # Using dtype as per previous correction
)
```

```
from peft import LoraConfig, get_peft_model, prepare_model_for_kbit_training

model = prepare_model_for_kbit_training(model)

lora_config = LoraConfig(
    r=8,
    lora_alpha=16,
    target_modules=["qkv_proj", "o_proj"],  # IMPORTANT for Phi-3
    lora_dropout=0.05,
    task_type="CAUSAL_LM"
)

model = get_peft_model(model, lora_config)
```

```
from transformers import TrainingArguments

training_args = TrainingArguments(
    output_dir="./phi3-legal",

    per_device_train_batch_size=1,
    gradient_accumulation_steps=8,

    num_train_epochs=1,  # start with 1

    learning_rate=2e-4,
    fp16=True,

    logging_steps=50,
    save_strategy="epoch",

    evaluation_strategy="steps",
    eval_steps=500,

    report_to="none"
)
```

```
from trl import SFTTrainer

trainer = SFTTrainer(
    model=model,
    train_dataset=train_dataset,
    eval_dataset=dev_dataset,
    dataset_text_field="text",
    tokenizer=tokenizer,
    args=training_args,
    max_seq_length=1024
)
```

```
trainer.train()
```

```
model.save_pretrained("./phi3-legal")
tokenizer.save_pretrained("./phi3-legal")
```

```
def generate_cot(case_text):
    prompt = f"""
You are an expert Indian legal assistant.

Case:
{case_text}

Think step-by-step:
1. Understand facts
2. Apply legal reasoning
3. Decide outcome

Final answer:
"""

    output = llm.create_chat_completion(
        messages=[
            {"role": "system", "content": "You are a legal expert."},
            {"role": "user", "content": prompt}
        ],
        max_tokens=300,
        temperature=0.7
    )

    return output["choices"][0]["message"]["content"]
```

```
from sentence_transformers import SentenceTransformer
import numpy as np

embedder = SentenceTransformer("all-MiniLM-L6-v2")

# build small index
texts = [" ".join(train_data[i]["text"]["facts-and-arguments"]) for i in range(3000)] # Extract and join text
embeddings = embedder.encode(texts)
```

```
def retrieve(query, k=3):
    q_emb = embedder.encode([query])
    scores = np.dot(embeddings, q_emb.T).squeeze()
    idx = np.argsort(scores)[-k:][::-1]
    return [texts[i] for i in idx]
```

```
def rag_cot(case_text):
    retrieved = retrieve(case_text, k=1)

    context = "\n\n".join(retrieved)

    prompt = f"""
You are a legal expert.

Similar Cases:
{context}

New Case:
{case_text}

Think step-by-step and give final decision.
"""

    output = llm.create_chat_completion(
        messages=[
            {"role": "system", "content": "Expert legal assistant"},
            {"role": "user", "content": prompt}
        ],
        max_tokens=300
    )

    return output["choices"][0]["message"]["content"]
```

```
sample_dict = train_data[0]["text"]
sample_text = " ".join(sample_dict["facts-and-arguments"]) # Extract and join relevant text
result = rag_cot(sample_text)

print(result)
```

```
def rag_cot_with_arguments(case_text):
    retrieved_cases = retrieve(case_text, k=1)

    context = "\n\n".join([
        f"Case {i+1}:\n{c}" for i, c in enumerate(retrieved_cases)
    ])

    prompt = f"""
You are an expert Indian legal assistant.

### Similar Cases:
{context}

### New Case:
{case_text}

### TASK:
Analyze the case and provide:

1. Key Facts
2. Arguments for the Petitioner
3. Arguments for the Respondent
4. Legal Reasoning (step-by-step)
5. Final Decision (Grant Bail / Reject Bail)

### Answer Format:

**Facts:**
...

**Petitioner Arguments:**
...

**Respondent Arguments:**
...

**Reasoning:**
...

**Final Decision:**
"""

    output = llm.create_chat_completion(
        messages=[
            {"role": "system", "content": "You are a legal expert who explains reasoning clearly."},
            {"role": "user", "content": prompt}
        ],
        max_tokens=500,
        temperature=0.6
    )

    return output["choices"][0]["message"]["content"]
```

```
sample_dict = train_data[0]["text"]
sample_text = " ".join(sample_dict["facts-and-arguments"]) # Extract and join relevant text

result = rag_cot_with_arguments(sample_text)

print(result)
```
